```python
import os
import json
from zipfile import ZipFile

import pandas as pd
import tensorflow as tf
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Embedding, LSTM
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
kaggle_ditionary = json.load(open('kaggle.json'))
```

```python
os.environ['KAGGLE_USERNAME'] = kaggle_ditionary['username']
os.environ['KAGGLE_KEY'] = kaggle_ditionary['key']
```

```python
data = pd.read_csv('imdb-dataset-of-50k-movie-reviews\\IMDB Dataset.csv')
```

```python
data['sentiment'].value_counts()
```

```python
data["sentiment"] = data["sentiment"].map({"positive": 1, "negative": 0}).astype("int32")
train_data, test_data = train_test_split(data, test_size=0.2, random_state=42)
```

```python
# Tokenizer config used for training artifacts (must match deployed model.pkl + tokenizer.pkl):
# num_words=5000, oov_token='<OOV>' so out-of-vocabulary words map to <OOV> instead of being dropped.
tokenizer = Tokenizer(num_words=5000, oov_token='<OOV>')
tokenizer.fit_on_texts(train_data["review"])
X_train = pad_sequences(tokenizer.texts_to_sequences(train_data["review"]), maxlen=200)
X_test = pad_sequences(tokenizer.texts_to_sequences(test_data["review"]), maxlen=200)
```

```python
Y_train = train_data["sentiment"]
Y_test = test_data["sentiment"]
```

```python
model = Sequential()
model.add(Embedding(input_dim=5000, output_dim=128, input_length = 200))
model.add(LSTM(128, dropout=0.2, recurrent_dropout=0.2))
model.add(Dense(1,activation='sigmoid'))
model.build(input_shape=(None, 200))
```

```python
model.summary()
```

```python

model.compile(optimizer='adam', loss="binary_crossentropy", metrics=["accuracy"])

# Max 15 epochs; stop early if val_loss does not improve for 2 epochs; keep best weights.
early_stopping = EarlyStopping(
    monitor='val_loss',
    patience=2,
    restore_best_weights=True,
)
history = model.fit(
    X_train,
    Y_train,
    epochs=15,
    batch_size=64,
    validation_split=0.2,
    callbacks=[early_stopping],
)
print(f"Stopped after {len(history.history['val_loss'])} epoch(s); "
      f"best val_loss={min(history.history['val_loss']):.4f}")
```

```python
loss, accuracy = model.evaluate(X_test,Y_test)
print(f"Test loss : {loss} Test accuracy : {accuracy}")
```

```python
def predict_sentiment(review):
    sequence = tokenizer.texts_to_sequences([review])
    padded_sequence = pad_sequences(sequence, maxlen = 200)
    prediction = model.predict(padded_sequence)
    sentiment = "positive" if prediction[0][0] > 0.5 else "negative"
    return sentiment
```

```python
new_review = "This movie was fantastic. I loved it."
sentiment = predict_sentiment(new_review)
print(f"The sentiment of the review is : {sentiment}")
```

```python
new_review = "This movie was not that good"
sentiment = predict_sentiment(new_review)
print(f"The sentiment of the review is: {sentiment}")
```
